## Last.Fm Features

```python
lfm = pd.read_csv('../data/raw/csv/lastfm_features.csv')

lfm['tags'] = lfm['tags'].fillna('')
lfm['similar_artists'] = lfm['similar_artists'].fillna('')

lfm['tags_list'] = lfm['tags'].apply(
    lambda x: [tag.strip().lower() for tag in x.split(',') if tag.strip()]
)

lfm['similar_artists_list'] = lfm['similar_artists'].apply(
    lambda x: [artist.strip() for artist in x.split(',') if artist.strip()]
)

lfm['tags_list'] = lfm.apply(
    lambda row: [tag for tag in row['tags_list'] if tag != row['artist'].lower()], 
    axis=1
)

lfm
```

## Candidate Pooling

```python
all_similar_artists = lfm['similar_artists_list'].explode().dropna()
unique_candidates = set(all_similar_artists)
known_artists = set(lfm['artist'])

unheard_candidates = unique_candidates - known_artists
print(f"Total known artists: {len(known_artists)}")
print(f"Total new candidate artists to query on Spotify: {len(unheard_candidates)}")
```
