## Notebook for Data Cleaning

### Candidate Tracks CSV

```python
import pandas as pd
import numpy as np

candidate_tracks = pd.read_csv("../data/processed/csv/candidate_tracks.csv")
candidate_tracks
```

```python
candidate_tracks.info()
```

```python
candidate_tracks.isna().sum()
print(candidate_tracks.duplicated().sum())
```

```python
candidate_tracks['release_date'] = pd.to_datetime(candidate_tracks['release_date'], errors='coerce', format='mixed')
candidate_tracks['duration_ms'] = candidate_tracks['duration_ms'].astype(np.float32)
candidate_tracks['duration_sec'] = candidate_tracks['duration_ms'] / 1000.0
candidate_tracks['duration_min'] = np.round(candidate_tracks['duration_ms'] / 60000.0, 2)

columns_arr = [
    'id', 'name', 'primary_artist_name', 'primary_artist_id', 'all_artists_featured',
    'album_name', 'release_date', 'duration_ms', 'duration_sec', 'duration_min', 'explicit'
]

candidate_tracks = candidate_tracks[[col for col in columns_arr if col in candidate_tracks.columns]]
candidate_tracks.info()
```

### Last.Fm Feats CSV

```python
lastfm = pd.read_csv('../data/processed/csv/lastfm_features.csv')
lastfm
```

```python
lastfm.info()
lastfm.isna().sum()
lastfm.duplicated().sum()
```

```python
lastfm['tags'] = lastfm['tags'].fillna('')
lastfm['similar_artists'] = lastfm['similar_artists'].fillna('')

lastfm['num_tags'] = lastfm['tags'].apply(lambda x: len(x.split(',')) if x else 0)
lastfm['num_similar_artists'] = lastfm['similar_artists'].apply(lambda x: len(x.split(',')) if x else 0)

lastfm.info()
```

```python
lastfm_features = lastfm.copy()
lastfm_features.head(10)
```

### Longest Top Tracks CSV

```python
ltt = pd.read_csv('../data/processed/csv/long_top_tracks.csv', index_col=0)
ltt
```

```python
ltt.isna().sum()
ltt.duplicated().sum()
```

```python
# helper function to transform the tracks csv

def transform_tracks(df: pd.DataFrame) -> pd.DataFrame:

    df['duration_ms'] = df['duration_ms'].astype(np.float32)
    df['explicit'] = df['explicit'].astype('int32')
    df['release_format'] = df['release_format'].astype(str).str.capitalize()
    df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce', format='mixed')
    df['duration_sec'] = df['duration_ms'] / 1000.0
    df['duration_min'] = np.round(df['duration_ms'] / 60000.0, 2)

    return df

long_top_tracks = transform_tracks(ltt)
long_top_tracks
```

### Medium Top Tracks

```python
mtt = pd.read_csv("../data/processed/csv/medium_top_tracks.csv", index_col=0)
mtt
```

```python
mtt.info()
mtt.duplicated().sum()
mtt.isna().sum()
```

```python
medium_top_tracks = transform_tracks(mtt)
medium_top_tracks
```

### Recently Played

```python
rp= pd.read_csv("../data/processed/csv/recently_played.csv", index_col=0)
rp
```

```python
rp.info()
```

```python
rp.duplicated().sum()
rp.isna().sum()
```

```python
rp.info()
```

```python
rp['played_at'] = pd.to_datetime(rp['played_at'], errors='coerce', format='mixed')
rp['release_date'] = pd.to_datetime(rp['played_at'], errors='coerce', format='mixed')
rp['release_format'] = rp['release_format'].astype(str).str.capitalize()
rp['duration_ms'] = rp['duration_ms'].astype(np.float32)
rp['duration_sec'] = rp['duration_ms'] / 1000.0
rp['duration_min'] = np.round(rp['duration_ms'] / 60000.0, 2)
rp['explicit'] = rp['explicit'].astype(int)

rp = rp.rename(columns={'context.type': 'context_type'})

rp['context_type'] = (
    rp['context_type']
    .rename('context_type')
    .fillna('artist')
    .astype(str).str.capitalize()
)

recently_played = rp.copy()
recently_played
```

## Saved Tracks

```python
st = pd.read_csv('../data/processed/csv/saved_tracks.csv', index_col=0)

saved_tracks = transform_tracks(st)
saved_tracks['added_at'] = pd.to_datetime(saved_tracks['added_at'], errors='coerce', format='mixed')
saved_tracks.info()
```

```python
saved_tracks.isna().sum()
print(saved_tracks.duplicated().sum())
```

### Short Top Tracks

```python
stt = pd.read_csv('../data/processed/csv/short_top_tracks.csv', index_col=0)
short_top_tracks = transform_tracks(stt)
short_top_tracks.info()
```

### User Playlists

```python
up = pd.read_csv('../data/processed/csv/user_playlists.csv', index_col=0)
up['collaborative'] = up['collaborative'].astype(int)
up['public'] = up['public'].astype(int)

up = up.dropna(subset=['name'])
up = up.rename(columns={'items.total' : 'total_items'})
up = up.drop(columns=['description'])

user_playlists = up.copy()
```

### Followed Artists

## Load the DataFrames

```python
to_csvs = [candidate_tracks, lastfm_features, long_top_tracks, medium_top_tracks, recently_played,
    saved_tracks, short_top_tracks, user_playlists, followed_artists]

filenames = ['candidate_tracks', 'lastfm_features', 'long_top_tracks', 'medium_top_tracks', 'recently_played',
    'saved_tracks', 'short_top_tracks', 'user_playlists', 'followed_artists']


for df, filename in zip(to_csvs, filenames):
    df.to_csv(f'../data/processed/clean_data/{filename}.csv')
```
